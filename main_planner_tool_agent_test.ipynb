```python
from command import CommandManager
from notion import Planner, Task
from agent import get_tool_agent
from pprint import pprint
from langchain_core.messages import SystemMessage, HumanMessage, ToolMessage


command_manager = CommandManager()
planner = Planner()
task_cls = Task

def format_messages(messages):
    for msg in messages:
        if isinstance(msg, HumanMessage):
            print(f"[Human]: {msg.content}")
        elif isinstance(msg, ToolMessage):
            print(f"[Tool - {msg.name}]: {msg.content}")
        elif hasattr(msg, 'tool_calls') and msg.tool_calls:
            # AIMessage with tool_calls
            print(f"[AI - Tool Calls]:")
            for tool_call in msg.tool_calls:
                if tool_call['args']:
                    print(f"  - {tool_call['name']} {tool_call['args']}")
                else:
                    print(f"  - {tool_call['name']}")
        elif hasattr(msg, 'content') and msg.content:
            # AIMessage with content
            print(f"[AI]")
            print(f"{msg.content}")
        elif hasattr(msg, 'name'):
            # ToolMessage
            print(f"[Tool - {msg.name}]: {msg.content}")
```

```python
agent = get_tool_agent(planner, task_cls)
```

```python
planner.load_tasks()
message = command_manager.execute_command("/show")
print(message)
```

```python
result = agent.invoke({"messages": [HumanMessage(content="2025년 11월 24일부터 2025년 11월 28일까지 '알바' 이름으로 12:00 ~ 18:00까지 '일상' 그룹으로 일정 5개 만들어")]}, {"thread_id": 1})
```

```python
format_messages(result['messages'])
```
